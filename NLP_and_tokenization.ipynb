What is Tokenization?

Tokenization is the process of breaking text into smaller units called tokens.
These tokens can be:

Words

Subwords

Characters

Sentences


Text: “NLP is amazing!”
Tokens → ["NLP", "is", "amazing", "!"]

# Why Do We Use Tokenization?
Because computers cannot understand raw text. They need text to be broken down into meaningful pieces before further processing like:

Text classification

Machine translation

Named entity recognition

Chatbots

LLMs

Tokenization makes text structured, analyzable, and convertible into numbers (which models understand).

# Types of Tokenization

1. Word Tokenization

Split text by words.

Example:
"He isn't playing" → ["He", "is", "n't", "playing"]

Problems:

Cannot handle unseen words

Vocab becomes huge

Sentence Tokenization

Break text into sentences.

Example:
"I live in Dubai. I study NLP."
→ ["I live in Dubai.", "I study NLP."]

Character Tokenization

Each character becomes a token.

"hello" → ["h", "e", "l", "l", "o"]

Used in:

OCR

Speech recognition

Some neural models

Subword Tokenization (Most Important for LLMs)

Break words into meaningful pieces (subwords).

Used in: BERT, GPT, RoBERTa, LLaMA, Gemini, etc.


Example: "playing" → ["play", "ing"]

```python
text = "NLP is amazing, isn't it?"
tokens = text.split()
print(tokens)
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

text = "NLP is powerful. It is used everywhere!"

# Word tokens
print(word_tokenize(text))

# Sentence tokens
print(sent_tokenize(text))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

text = "ChatGPT is transforming NLP."

doc = nlp(text)

for token in doc:
    print(token.text)
```
