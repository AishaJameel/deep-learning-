Remove HTML Tags

```python
text2 = "<p>Hello</p> <b>World</b>"
```

```python
import re
clean2 = re.sub(r'<.*?>', '', text2)
print(clean2)
```

Remove URLs

```python
text3 = "Visit https://google.com now!"
```

```python
clean3 = re.sub(r'http\S+|www\S+|https\S+', '', text3)
print(clean3)
```

Remove Mentions (@username)

```python
text4 = "Hey @john, meet @sara!"
```

```python
clean4 = re.sub(r'@\w+', '', text4)
print(clean4)
```

Remove Hashtags

```python
text5 = "Loving this! #Awesome #FunDay"
```

```python
clean5 = re.sub(r'#', '', text5)
print(clean5)
```

Remove Numbers

```python
text6 = "I have 2 dogs and 3 cats."
```

```python
clean6 = re.sub(r'\d+', '', text6)
print(clean6)
```

Remove Punctuation

```python
text7 = "Hello!!! How are you?? :))"

```

```python
clean7 = re.sub(r'[^\w\s]', '', text7)
print(clean7)
```

Remove Special Characters

```python
text8 = "Hello *&^% World!!!"
```

```python
clean8 = re.sub(r'[^a-zA-Z0-9\s]', '', text8)
print(clean8)
```

Remove Emojis

```python
text9 = "I love NLP 😍🔥😂"
```

```python
emoji_pattern = re.compile(
    "["
    u"\U0001F600-\U0001F64F"
    u"\U0001F300-\U0001F5FF"
    u"\U0001F680-\U0001F6FF"
    u"\U0001F1E0-\U0001F1FF"
    "]+", flags=re.UNICODE)

clean9 = emoji_pattern.sub(r'', text9)
print(clean9)
```

Remove Extra Spaces

```python
text10 = "This   has   too    many   spaces!"
```

```python
clean10 = re.sub(r'\s+', ' ', text10).strip()
print(clean10)
```

Remove Repeated Characters ("heyyyy" → "hey")

```python
text11 = "heyyyyy booooyyyy!!!!"
```

```python
clean11 = re.sub(r'(.)\1{2,}', r'\1', text11)
print(clean11)
```

Expand Contractions (don’t → do not)

```python
text12 = "I can't do this, it's not fair!"
```

```python
contractions = {
    "can't": "cannot",
    "don't": "do not",
    "it's": "it is",
    "i'm": "i am"
}

pattern = re.compile('|'.join(contractions.keys()))
clean12 = pattern.sub(lambda x: contractions[x.group()], text12.lower())
print(clean12)
```

Stopword Removal

```python
tokens14 = ["this", "is", "a", "good", "day"]
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

clean14 = [w for w in tokens14 if w not in stop_words]
print(clean14)
```
