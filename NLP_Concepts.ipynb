stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["playing", "studies"]

for w in words:
    print(w, "→", stemmer.stem(w))
```

Lemmatization

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemm = WordNetLemmatizer()
words = ["studies", "playing"]

for w in words:
    print(w, "→", lemm.lemmatize(w))
```

Bag of Words (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["Dog bites man", "Man bites dog"]

cv = CountVectorizer()
vectors = cv.fit_transform(docs)

print(cv.get_feature_names_out())
print(vectors.toarray())
```

High dimensionality (huge sparse vectors)

No meaning — “good” and “nice” are unrelated

No context — “Apple” (fruit) vs “Apple” (company)

Order of words lost

Fails badly with large datasets

```python
from gensim.models import Word2Vec

sentences = [
    ["dog", "bites", "man"],
    ["man", "bites", "dog"],
    ["dog", "eats", "food"],
    ["cat", "eats", "food"]
]

model = Word2Vec(sentences, min_count=1, vector_size=10)

print(model.wv["dog"])  # word vector for 'dog'
print(model.wv.most_similar("dog"))
```
